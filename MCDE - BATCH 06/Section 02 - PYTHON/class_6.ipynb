```python

class EmailNotification:
    def send(self, message):
        print(f"Email sent: {message}")

class SMSNotification:
    def send(self, message):
        print(f"SMS Sent: {message}")

class PushNotification:
    def send(self, message):
        print(f"Push Notification: {message}")

# generic method
def send_notification(notification, message):
    notification.send(message)


# Pass different notification objects to the same function
email = EmailNotification()
sms = SMSNotification()
push = PushNotification()

send_notification(email, "Your order is confirmed.")
send_notification(sms, "Your order is confirmed.")
send_notification(push, "Your order is confirmed.")


# Add a WhatsAppNotification class without modifying send_notification()
class WhatsAppNotification:
    def send(self, message):
        print(f"WhatsApp Sent: {message}")

whatsapp = WhatsAppNotification()
send_notification(whatsapp, "Your order is confirmed.")
```

### Encapsulation :
process of hiding details


```python
class saylani():
    def __init__(self,name,email,cnic):
        self.name = name # public
        self._email = email # protected
        self.__cnic = cnic  # private

    def public_method(self):
        print("this is public method")

    def _protected_method(self):
        print("this is protected method")

    def __private_method(self):
        print("this is private method")
```

```python
obj = saylani("Qasim", "qasim@gmail.com", "43211-98363654-2")
```

```python
print(obj.name)
print(obj._email)
print(obj._saylani__cnic)
```

```python
obj.public_method()
obj._protected_method()
obj._saylani__private_method()
```

### Task
- use encapsulation for class i.e StudentLogin():
- initialize two static but private attributes __user_name,__password
- create one private methods i.e name __dbconnectivity() that will validate you username and passward.if both are valid then should print "valid user" otherwise "not valid"
- create one public method i.e update passwaord and it should update your private password

```python
class StudentLogin():
    def __init__(self):
        self.__user_name = "Username"
        self.__password = "password"

    def __dbconnectivity(self, username, password):
        if self.__user_name == username and self.__password == password:
            print(f"valid user")
        else:
            print(f"Invalid user")

    def update_password(self, username, password, new_pass):
        if self.__user_name == username and self.__password == password:
            self.__password = new_pass
            print(f"passwaord updated...!")
        else:
            print(f"Invalid user")
```
