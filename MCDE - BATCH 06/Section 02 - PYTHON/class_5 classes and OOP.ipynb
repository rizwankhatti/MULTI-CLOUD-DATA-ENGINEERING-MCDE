### Human

- attributes/characteristics
  - name
  - color
  - weight
  - height
  - gender

- methods/actions/behaviour
  - breathing
  - sleeping
  - working

```python
class human:
    # adding class variable
    city = "Karachi"
    country = "Pakistan"
    def __init__(self,name,color,weight,height,gender):
        self.name = name
        self.color = color
        self.weight= weight
        self.height = height
        self.gender = gender

    def breathing(self):
        print(f"{self.name} is taking a deep breath")

    def sleeping(self):
        print(f"{self.name} is fast asleep and dreaming")

    def working(self):
        print(f"{self.name} is focused and hard at work")

```

```python

qasim = human("Qasim", 'brown', 70, 5.9, 'male')
ayesha = human("Ayesha", 'brown', 50, 5.2, 'female')
```

```python

# calling all arrtibutes and methods for qasim
print(qasim.name)
print(qasim.color)
print(qasim.weight)
print(qasim.height)
print(qasim.gender)
qasim.breathing('calmly')
qasim.sleeping(5)
qasim.working()

## calling class variables
print(qasim.city)
print(qasim.country)
```

```python
# calling all arrtibutes and methods for ayesha
print(ayesha.name)
print(ayesha.color)
print(ayesha.weight)
print(ayesha.height)
print(ayesha.gender)
ayesha.breathing('normally')
ayesha.sleeping(8)
ayesha.working()

# calling class variable
print(ayesha.city)
print(ayesha.country)
```

```python
person1 = human(
    name = "Rizwan",
    color = "brown",
    weight = 64,
    height = 590,
    gender = "male"
)

# accessing attributes
print(f"name: {person1.name}")
print(f"gender: {person1.gender}")

# calling methods
person1.breathing()
person1.working()
person1.sleeping()
```

```python
class CEO:

  def __init__(self, login, password):
    self.login = login
    self.password = password

  def Project(self,id, taskId, access, name):
    print(f"Project id is {id}")
    print(f"Project name is {name}")
    print(f"Project task id is {taskId}")
    print(f"Project access is {access}")

  def Report(self,time, date, task, name):
    print(f"Project name is {name}")
    print(f"Report date {date}")
    print(f"Report time {time}")
    print(f"Task name is {task}")

  def Employee(self, id):
    print(f"Employee login is {self.login}")
    print(f"Employee password is {self.password}")
```

# Example of inheritance

```python


# Example of Inhertance

class parents():

  def __init__(self, eye_color, skin_color, blood_group):
    self.eye_color = eye_color
    self.skin_color = skin_color
    self.blood_group = blood_group

  def hobby(self, hobby):
    print(f"have {hobby} Hobby")
        
```

```python


class child(parents):

  def __init__(self, eye_color, skin_color, blood_group, name):
    super().__init__(eye_color, skin_color, blood_group)
    self.name = name

  def coding(self):
    print(f"{self.name} Know Programing")
     
```

```python

pobj = parents('Black', "brown", 'A+')

## calling attribute and method of parent class
print(pobj.blood_group)
pobj.hobby('Newspaper Reading')
```

```python

cobj = child('Brown', 'black', 'AB+', 'Ali')

## calling attribute and method of child class
print(cobj.blood_group)
cobj.hobby('Reading')
cobj.coding()
```
