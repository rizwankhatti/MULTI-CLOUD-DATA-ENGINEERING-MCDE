```python
lst1 = [
    ['Qasim','Ali','Mehmood','Uzair','Ayan'],
    ['Hassan','Raza','Khan','Shah','Hussain'],
    [20,21,22,23,24],
    [40,41,42,43,44]
]
```

```python
lst1
```

```python
lst1[2]
```

```python
lst1[2][4]
```

```python
lst1[0][4], lst1[1][4], lst1[2][4], lst1[3][4]
```

```python
list((lst1[0][4], lst1[1][4], lst1[2][4], lst1[3][4]))
```

```python
lst2 = [
    ['Qasim','Ali','Mehmood','Uzair','Ayan'],
    ('Hassan','Raza','Khan','Shah','Hussain'),
    set({20,21,22,23,24}),
    [40,41,42,43,44]
]
```

```python
lst2
```

```python
lst2[0][4], lst2[1][4], list(lst2[2])[4], lst2[3][4] # list and tuple slicing are same,, set ki slici
ng mai type cast krna parta hai phele
```

```python
dict2 = {
   'f_name' : ['Qasim','Ali','Mehmood','Uzair','Ayan'],
   'l_name' : ('Hassan','Raza','Khan','Shah','Hussain'),
   'age' :  set({20,21,22,23,24}),
   'weight' :  [40,41,42,43,44]
}
```

```python
dict2['f_name'][4], dict2['l_name'][4], list(dict2['age'])[4], dict2['weight'][4]
```
