## Recap Functions

```python
def details(name): # name ==> parameter
    return name
```

```python
details("Rizwan") # value ==> argument
```

- args*
- kwargs**

```python
def details(*args):
    value = args
    return value # return tuple
```

```python
details("Goagle", "Python", "Sql")
```

```python
def details(**kwargs):
    value = kwargs
    return value # return dictionary
```

```python
details(company= "Goagle", course = "Python") # no need to quatation to key but quatation need to value
```

```python
def details(company, *args, **kwargs):
    print(company)
    print(args)
    print(kwargs)
```

```python
details("Goagle", "python","sql","sql", name = "Rizwan", course = "MCDE", city = "karachi")
```

```python
def details(company, *args, **kwargs):
    print(f"company {company}")
    print(f"args {args}")
    print(f"kwargs {kwargs}")
```

```python
details("Goagle", "python","sql","sql", name = "Rizwan", course = "MCDE", city = "karachi")
```
