#### Q2. List Slicing
numbers = [5, 10, 15, 20, 25, 30, 35, 40, 45, 50]

Create three new lists: first five elements, last four elements, and every second element in reverse order

```python
numbers = [5, 10, 15, 20, 25, 30, 35, 40, 45, 50]
```

```python
first_five = numbers[:5]
last_four = numbers[-4:]
every_second_reverse = numbers[::-2]
```

```python
print("First five elements : ", first_five)
print("Last four elements : ", last_four)
print("every second element in reverse order", every_second_reverse)
```

```python
numbers
```
